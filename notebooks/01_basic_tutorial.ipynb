# Get data

Select a symbol, a tick interval, and a time interval -- or just a limit of candles.

Time interval inputs can be a string like `2022-02-14 12:00:00`.

```python
# Fetch the last 200 candles for LUNCUSDT, 1-hour interval
lun = binpan.Symbol(symbol='luncusdt',
                    tick_interval='1h',
                    limit=200,
                    time_zone='Europe/Madrid',
                    closed=True)
lun.df
```

```python
list(lun.df.columns)
```

## Repair kline discontinuity

If kline data has gaps (e.g. from the exchange not returning certain candles), `repair_continuity()` will fill them with the previous candle's Close value.

```python
# Check and repair any gaps in the kline data
lun.repair_continuity()
```

## Several indicators in one figure

BinPan places overlays (EMA, Bollinger...) on the candles and oscillators (MACD, RSI...) in their own panels automatically. Add several and `plot()` lays them out together.

```python
dash = binpan.Symbol(symbol='luncusdt', tick_interval='1h', limit=200, time_zone='Europe/Madrid', closed=True)
dash.ema(12, color='blue')
dash.ema(26, color='orange')
dash.bbands()
dash.macd()
dash.rsi()
dash.plot(title='EMA + Bollinger + MACD + RSI in one figure')
```

## Ichimoku cloud

One of the more visual indicators: the Ichimoku *Kumo* cloud (Tenkan, Kijun, Senkou A/B and Chikou span), plotted directly over the candles.

```python
ichi = binpan.Symbol(symbol='luncusdt', tick_interval='1h', limit=200, time_zone='Europe/Madrid', closed=True)
ichi.ichimoku()
ichi.plot(title='LUNCUSDT — Ichimoku cloud')
```

# Trades

Trades can be downloaded for the time period of the instance.

**Beware of requesting large time periods.**

BinPan monitors the API weight and waits if it is approaching the limit.

```python
lun.get_atomic_trades(minutes=120)
```

```python
lun.get_agg_trades(minutes=60)
```

# Plots with trades

Using `plot_agg_trades_size()`, you can visualize the size of aggregated trades. Useful for finding whale operations or support/resistance levels.

```python
lun.plot_agg_trades_size()
```

## Resample klines

You can resample klines to a different (larger) timeframe. It only aggregates **upwards** (e.g. 1h candles into 4h); resampling to a smaller interval is not possible.

```python
resampled = lun.resample('4h')
resampled
```

## Drop indicators

```python
lun.drop()
list(lun.df.columns)
```

## Heikin-Ashi candles

BinPan can compute Heikin-Ashi candles from normal OHLCV data. These smooth out price action and make trends easier to identify.

```python
lun.hk(inplace=True)
lun.plot()
```
